```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

print("NUMPY")
arr = np.array([1, 2, 3, 4, 5])
print("1D Array:")
print(arr)
print(type(arr))
print(arr.shape)

arr2 = np.array([[1, 2], [3, 4]])
print("\n2D Array:")
print(arr2)
print(arr2.shape)

arr3 = np.array([
    [[1, 2], [3, 4]],
    [[5, 6], [7, 8]]
])

print("\n3D Array:")
print(arr3)
print(arr3.shape)

print("\nList vs Array")

py_list = [1, 2, 3]
print("Python List * 2:")
print(py_list * 2)

np_array = np.array([1, 2, 3])
print("NumPy Array * 2:")
print(np_array * 2)

print("\nPANDAS")

data = {
    "Name": ["Asha", "Rahul", "Neha"],
    "Marks": [85, 90, 78]
}

df = pd.DataFrame(data)

print("\nStudent DataFrame:")
print(df)

data_dict = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eva'],
    'Age': [25, np.nan, 30, 22, 28],
    'Score': [85, 90, 95, 80, 70]
}

df2 = pd.DataFrame(data_dict)

print("\nDataset with Missing Values:")
print(df2)

print("\nNull Values:")
print(df2.isnull().sum())

df2['Age'] = df2['Age'].fillna(df2['Age'].mean())

print("\nAfter Filling Missing Values:")
print(df2)

print("\nMATPLOTLIB")

x = [1, 2, 3, 4, 5]
y = [10, 14, 12, 18, 20]

plt.plot(x, y, marker='o')

plt.title("Simple Line Chart")
plt.xlabel("Time")
plt.ylabel("Distance")

plt.grid(True)

plt.show()

students = ["A", "B", "C", "D", "E"]
marks = [85, 70, 90, 65, 95]

plt.bar(students, marks)

plt.title("Student Marks")
plt.xlabel("Students")
plt.ylabel("Marks")

plt.show()

activities = ["Study", "Exercise", "Sleep", "Hobbies"]
hours = [4, 1, 7, 2]

plt.pie(hours, labels=activities, autopct='%1.1f%%')

plt.title("Daily Activities")

plt.show()

data = [10, 20, 20, 30, 40, 40, 40, 50]

plt.hist(data)

plt.title("Histogram")

plt.show()

x = [1, 2, 3, 4, 5]
y = [5, 7, 8, 10, 12]

plt.scatter(x, y)

plt.title("Scatter Plot")

plt.show()
```
